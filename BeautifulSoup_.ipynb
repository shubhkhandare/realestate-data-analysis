```python
from bs4 import BeautifulSoup
```

```python
import requests
```

```python
respose = requests.get("https://scrapifydatalabs.com/playground/nestly/")
```

```python
print(respose)
```

```python
soup = BeautifulSoup(respose.text, "html.parser")
```

```python
print(soup.title.text)
```

```python
home = soup.find("div", id="nl-home-list")
```

```python
print(home)
```

```python
card = soup.find_all("article", class_="nl-card")
```

```python
print(card)
```

```python
import pandas as pd

all_prop_data = []

for i in card:
  price = i.find("p", class_="nl-price").text
  fact = i.find("p", class_="nl-facts").text
  meta = i.find("div", class_="nl-meta").text
  broker = i.find("div", class_="nl-broker").text
  #print(price, "|", fact, "|", meta, "|", broker)
  all_prop_data.append([price,fact,meta,broker])

df = pd.DataFrame(all_prop_data, columns=['price','fact','meta','broker'])
```

```python
display(df)
```

```python
print(df["fact"].head())
```

```python
df[["bedroom", "bathroom", "sqft", "property_type"]] = df["fact"].str.split("|", expand=True)
```

```python
display(df)
```

```python
df["sqft"] = (
    df["sqft"]
    .str.replace(" sqft", "")
    .str.replace(",", "")
    .astype(int)
)
```

```python
print(df["sqft"])
```

```python
print(df["price"].unique())
```

```python
df["listing_type"] = df["price"].apply( lambda x: "Rent" if "/mo" in x else "Sale")
```

```python
print(df[["price", "listing_type"]])
```

```python
df["price"] = df['price'].str.replace("$", '').str.replace(",", '').str.replace("/mo", '').str.strip().astype(int)
```

```python
display(df)
```
