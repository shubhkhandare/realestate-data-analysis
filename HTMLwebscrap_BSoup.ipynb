```python
html = """
<html>
<body>

<div class="property">
    <h2 class="property-name">3 BHK Flat</h2>
    <span class="location">Balewadi</span>
    <span class="budget">60 Lakh</span>
</div>

<div class="property">
    <h2 class="property-name">2 BHK Flat</h2>
    <span class="location">Baner</span>
    <span class="budget">45 Lakh</span>
</div>

<div class="property">
    <h2 class="property-name">4 BHK Villa</h2>
    <span class="location">Wakad</span>
    <span class="budget">1 Crore</span>
</div>

</body>
</html>
"""
```

```python
from bs4 import BeautifulSoup
```

```python
soup = BeautifulSoup(html, 'html.parser')
```

###I need property name, location and budget for every listing.”

```python
result = soup.find_all("div", class_ = "property")
```

```python
for i  in result:
  property_name = i.find("h2", class_="property-name").text
  location = i.find("span", class_="location").text
  budget = i.find("span", class_="budget").text
  print(property_name, location, budget)
```

```python
import pandas as pd

properties_data = []
for i in result:
    property_name = i.find("h2", class_="property-name").text
    location = i.find("span", class_="location").text
    budget = i.find("span", class_="budget").text
    properties_data.append([property_name, location, budget])

df_prop = pd.DataFrame(properties_data, columns= ['name', 'location', 'budget'])
display(df_prop)
```

```python
def converter(value):

  if isinstance(value, (int, float)):
      return value


  value = str(value).strip()

  if 'Crore' in value:
    number = float(value.replace('Crore', '').strip())
    return number * 10000000

  elif 'Lakh' in value:
    number1 = float(value.replace('Lakh', '').strip())
    return number1 * 100000
  else:
    try:
       return float(value)
    except ValueError:
       return None
```

```python
df_prop['budget'] = df_prop['budget'].apply(converter)
```

```python
df_prop.head()
```

```python
df_prop.drop_duplicates(subset= ['location', 'budget'])
```

##Now give me only properties where the budget is greater than ₹50 lakh, and show the result sorted from highest budget to lowest

```python
df_prop[df_prop['budget'] > 5000000].sort_values(by='budget', ascending=False)
```

##save

```python
df_prop = df_prop.sort_values(by='budget', ascending=False)
```
